```python
# --------------------------------------------
#           UNIVARITATE ANALYSIS
# --------------------------------------------
```

```python
# Survival Distribution
ax = sns.countplot(x='Survived', data=df)

ax.bar_label(ax.containers[0])

plt.title('How Many Passengers Survived?')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# --> Observations :
# 549 passengers did not survive.
# 342 passengers survived.
```

```python
# Gender Distribution
ax = sns.countplot(x='Sex', data=df)
ax.bar_label(ax.containers[0])
plt.title('What is the gender distribution of passengers?')
plt.show()
```

```python
# --> Observations :
# There were 577 male passengers and 314 female passengers.
# Male passengers significantly outnumbered female passengers.
```

```python
# Passenger Class Distribution
ax = sns.countplot(x='Pclass', data=df)
ax.bar_label(ax.containers[0])
sns.countplot(x='Pclass', data=df)
plt.title('How are passengers distributed across classes?')
plt.show()
```

```python
# --> Observations :
# Third Class (491 passengers) had the highest number of passengers.
# First Class had 216 passengers.
# Second Class had 184 passengers.
```

```python
# Age Distribution
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('What is the age distribution of passengers?')
plt.show()
```

```python
# --> Observations :
# Most passengers were between 20 and 40 years old.
# The peak age is around 30 years.
# Very few passengers were above 60 years old.
```

```python
# Fare Distribution
sns.histplot(df['Fare'], bins=20, kde=True)
plt.title('How are ticket fares distributed?')
plt.show()
```

```python
# --> Observations :
# Most passengers paid low ticket fares, generally below 50.
# A small number of passengers paid very high fares, creating a long right tail.
```

```python
# Key Findings from Univariate Analysis : 
# Most passengers were aged between 20 and 40 years.
# The majority of passengers traveled in Third Class.
# Male passengers outnumbered female passengers.
# Most ticket fares were relatively low, with only a few expensive tickets.
# More passengers died than survived during the Titanic disaster.
```

```python
# ------------------------------------
#         BIVARITAE ANALYSIS
# ------------------------------------
```

```python
# Survival vs Gender
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Did gender influence survival?')
plt.show()
```

```python
# Observation:

# Female passengers had a much higher survival rate than male passengers.
# Most male passengers did not survive, while the majority of female passengers survived.
# This suggests that gender was one of the strongest factors affecting survival.

# Finding:

# Female passengers were significantly more likely to survive than male passengers, indicating priority evacuation for women.
```

```python
# Survival vs Passenger Class
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Did passenger class affect survival?')
plt.show()
```

```python
# Observation:

# First-class passengers had more survivors than non-survivors.
# Third-class passengers had the highest number of deaths.
# Second-class passengers showed a more balanced survival pattern.

# Finding:

# Passenger class strongly influenced survival, with first-class passengers having the best survival chances and third-class passengers the lowest.
```

```python
# Survival vs Embarked
sns.countplot(x='Embarked', hue='Survived', data=df)
plt.title('Did port of embarkation affect survival?')
plt.show()
```

```python
# Observation:

# Most passengers embarked from Southampton (S), and deaths exceeded survivals there.
# Passengers embarking from Cherbourg (C) had relatively high survival counts.
# Queenstown (Q) had the fewest passengers and survivors.

# Finding:

# Survival rates varied across embarkation ports, with passengers from Cherbourg showing better survival outcomes than those from Southampton.
```

```python
# Age vs Survival
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Did age impact survival?')
plt.show()
```

```python
# Observation:

# The median age of survivors and non-survivors is very similar (around 28–30 years).
# Survivors include both young children and older adults.
# Age distributions overlap considerably between the two groups.
```

```python
# Fare vs Survival
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title('Did higher fare passengers survive more often?')
plt.show()
```

```python
# Observation:

# Survivors generally paid higher ticket fares than non-survivors.
# The median fare for survivors is noticeably higher.
# Several high-fare passengers survived, creating higher-value outliers in the survivor group.
```

```python
# Overall EDA Conclusion
# 549 passengers did not survive, while 342 survived.
# Female passengers had a significantly higher survival rate than males.
# First-class passengers were more likely to survive than third-class passengers.
# Most passengers were aged between 20 and 40 years.
# Higher ticket fares were associated with higher survival rates.
# Cherbourg passengers showed relatively better survival outcomes.
# Passenger class and gender were the strongest factors influencing survival.
# Age had a comparatively weaker impact on survival outcomes.
```

```python
# -------------------------------------
#         Correlation Analysis
# -------------------------------------
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Observation:

# Passenger class (Pclass) has a moderate negative correlation with Survival (-0.34), meaning higher-class passengers were more likely to survive.
# Fare has a positive correlation with Survival (0.26), indicating that passengers paying higher fares tended to survive more often.
# Fare and Pclass show a strong negative correlation (-0.55), meaning higher-class passengers generally paid higher fares.
# SibSp and Parch have a moderate positive correlation (0.41), suggesting family members often traveled together.
# Age has very weak correlation with Survival (-0.07).

# Finding:

# Passenger class and fare were the most influential factors associated with survival, while age showed only a weak relationship with survival outcomes.
```
